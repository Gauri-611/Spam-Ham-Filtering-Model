```python
# Email Spam Detection using Machine Learning
# Algorithm: Naive Bayes

# Step 1: Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Step 2: Create dataset
data = {
    "email": [
        "Congratulations! You won a free lottery prize",
        "Claim your free money now",
        "You have won a cash prize",
        "Click here to get your free reward",
        "Win a free iPhone today",
        "You are selected for a free gift",
        "Get rich quickly by clicking this link",
        "Congratulations you won a lottery",

        "Hello, how are you?",
        "Please send me the assignment",
        "The meeting is scheduled for tomorrow",
        "Can you send me the notes?",
        "Your class starts at 10 AM",
        "Please call me when you are free",
        "Let's meet tomorrow",
        "Here is the document you asked for"
    ],

    "label": [
        "spam", "spam", "spam", "spam",
        "spam", "spam", "spam", "spam",
        "ham", "ham", "ham", "ham",
        "ham", "ham", "ham", "ham"
    ]
}

# Step 3: Convert data into DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Step 4: Separate input and output
X = df["email"]
y = df["label"]

# Step 5: Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 6: Convert text into numerical features
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Step 7: Create Naive Bayes model
model = MultinomialNB()

# Step 8: Train the model
model.fit(X_train_tfidf, y_train)

# Step 9: Make predictions
y_pred = model.predict(X_test_tfidf)

# Step 10: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

# Step 11: Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 12: Test a new email
new_email = input("\nEnter an email message: ")

new_email_tfidf = vectorizer.transform([new_email])

prediction = model.predict(new_email_tfidf)

print("\nResult:", prediction[0].upper())

if prediction[0] == "spam":
    print("This is a SPAM email.")
else:
    print("This is NOT a SPAM email.")
```
